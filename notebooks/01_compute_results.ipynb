# Stage 1: compute and store results

This notebook writes numerical outputs into `cached_results/`. Stage 2 (`02_reproduce_figures.ipynb`) loads those files and draws the paper figures.

Cheap experiments (teaser + Task 1) run on CPU. Task 2 and the GPU experiments already have stored per-replication CSVs in `cached_results/`. Do not rerun those jobs to reproduce the paper figures.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from picpi.paths import cached
from picpi.teaser import compute_teaser, save_teaser
from picpi.task1 import (
    compute_misspecified_tree,
    compute_multivariate_table,
    compute_univariate_visualization,
    save_task1,
)

print(ROOT)
```

## Teaser (`fig:teaser`)

Quadratic ground truth, linear logistic fit, empirical PICPI bins. Writes `cached_results/teaser/`.

```python
teaser = compute_teaser()
teaser_dir = save_teaser(teaser)
print(teaser_dir)
teaser["intervals"]
```

## Task 1 visualization, tree figure, and Monte Carlo table

Writes `cached_results/task1/`. The 100-replication table takes about two minutes. Set `N_MC = 2` for a smoke test.

```python
N_MC = 100

univariate = compute_univariate_visualization()
tree = compute_misspecified_tree()
mc_results = compute_multivariate_table(n_mc=N_MC)
task1_dir = save_task1(univariate, tree, mc_results)
print(task1_dir)
mc_results.groupby("Method")[["Average length", "ECE", "Coverage p*"]].mean()
```
